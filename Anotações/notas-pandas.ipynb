```python
# Bibliotecas que vou usar
import pandas as pd
```

## **Pandas:** facilita manipulação de dados tabulares, tipo CSV, Excel, Parquet (inclusive pode ser usado para entrada de dados);

- O pandas utiliza o pŕoprio *numpy* por trás pra representar *estruturas e tipos de dados*;
- Tipos de dados normais: int, float e bool. Tipos *compostos* são representados pelo tipo **"object"**, tipo listas, dicionários e **strings**;
- Datas podem ser representadas usando **datetime64**;
- O tipo **category** pode ser usado para guardar categorias que se repetem com alta frequência (ex: Notas de 1 a 10) de forma otimizada em gasto de memória;

### **Series:** são um tipo fundamental de dado no Pandas, e representam **sequências unidimensionais** de dados;
- Compostas de índices/index que são usados para acessar cada dado da Series, que é geralmente um número inteiro ou uma string;

### **Dataframes:** são a classe mais utilizada no Pandas, e representam tabelas com várias colunas e linhas;
- **Index:** indexa as linhas do DF. Geralmente númerico, mas pode ser composto por strings;
- **Colunas:** cada coluna do DF é uma Series com certos valores;

### **#** Atributos e métodos gerais dos DFs

**Atributos:**
- **df.shape:** retorna uma Tupla, cujo primeiro elemento é o n° de linhas, e o segundo o n° de colunas do dataframe;
- **df.index:** objeto do index de um df;
- **df.columns:** objeto que contém as colunas do df;

**Métodos gerais:**
- **df.head(n):** retorna as n primeiras linhas do DF, por padrão n=5;
- **df.info()**: printa no terminal um sumário de características do df (os index, o consumo de memória, as colunas, número de dados não nulos, o tipo de dado);

**Acesso de colunas:**

```python
Series_da_coluna = df["Country or region"] # Retorna um Series
df_duas_colunas = df[["Country or region", "Perceptions of corruption"]] # Retorna um dataframe, serve para acessar 2 ou mais colunas com uma lista.

print(Series_da_coluna)
print()
print(df_duas_colunas)
```

**Acesso de linhas:**

```python
# .loc acessa uma linha pelo valor de índice;
linha_a = df.loc[0]
linha_b = df.iloc[3] # .iloc seleciona pela posição, mesmo que o índice use strings


# Slicing de várias linhas do DF:
df_linhas_a = df.loc[1:3]
# no .iloc o limite final é exclusivo; no .loc, ele é inclusivo.

# Para adicionar ou atualizar uma linha, atribua uma Series com os mesmos rótulos:
# df.loc[156] = {"Country or region": "Exemplo", "Score": 0.0}

print(linha_a)
print(linha_b)
print(df_linhas_a)
```

**Acesso de valores individuais:**

```python
# .at acessa valores individuais baseado no valor do índice e da coluna;
val_a = df.at[0, "Perceptions of corruption"]
print(val_a)

# .iat é igual .iloc, usa a posição numérica de colunas e linhas.

val_b = df.iat[0, 8]
print(val_a == val_b)

# Pode atualizar os valores com isso também igual um vetor.
```

### **# DF view X copy**
Como DFs podem conter **muitos dados e usar bastante memória**, a maioria das operações que envolvem essa classe usam passagem por referência e diferentes views de um mesmo objeto em memória.

```python
view_df = df[["GDP per capita", "Country or region"]]
print(view_df.head())
# O retorno de uma seleção pode ser view ou cópia; não dependa desse detalhe.

copy_df = df[["GDP per capita", "Country or region"]].copy()
copy_df.at[0, "GDP per capita"] = 2.000
print("Valor na cópia:", copy_df.at[0, "GDP per capita"])
print("Valor no original:", df.at[0, "GDP per capita"])

# Uma copy modifica um valor somente na cópia, e não modifica o df original.
```

### **# Métodos para Series/Colunas**
Para **análise, transformação e filtragem de dados**, é possível utilizar métodos de Series aplicados nas colunas de uma DF.

**Métodos estatísticos:**

```python
estatisticas = {
    "soma": df["GDP per capita"].sum(),
    "média": df["GDP per capita"].mean(),
    "mediana": df["GDP per capita"].median(),
    "mínimo": df["GDP per capita"].min(),
    "máximo": df["GDP per capita"].max(),
    "variância": df["GDP per capita"].var(),
    "desvio-padrão": df["GDP per capita"].std(),
}
print(pd.Series(estatisticas))
print("Moda:\n", df["GDP per capita"].mode())

# Retorna o número de elementos únicos na coluna
print(df["Country or region"].nunique())
```

### **# Merge/Join**
O `merge` do pandas combina linhas de dois DataFrames usando uma ou mais colunas-chave. O parâmetro `how` define quais chaves serão mantidas.


```python
df_esq = df[["Country or region", "Score"]].head(10)
df_dir = df[["Country or region", "GDP per capita"]].head(10)
novo_df = df_esq.merge(
    df_dir,
    how="inner",
    on="Country or region",
    sort=True,
)
print(novo_df)

# Quando as chaves têm nomes diferentes, use left_on e right_on no lugar de on:
# novo_df = df_esq.merge(df_dir, left_on="pais", right_on="nome", how="left")
```

### **# GroupBy/Agrupamentos**
**Funções de agrupamento:** agregam as colunas que não foram usadas como chave, retornando um resumo por grupo.



```python
faixa_gdp = pd.cut(
    df["GDP per capita"],
    bins=3,
    labels=["baixo", "médio", "alto"],
)
score_por_faixa = df.groupby(faixa_gdp, observed=True)["Score"].mean()

print(score_por_faixa)
```
