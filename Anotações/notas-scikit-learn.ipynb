## **O que é o scikit-learn?**
- É uma biblioteca do Python para aprendizado de máquina (machine learning).
- Ela é capaz de trabalhar com diversos modelos, sem que o programador se preocupe com o funcionamento matemático interno.

### **# O que é machine learning?**
- A partir de dados tratados e informações estatísticas que obtemos deles, queremos fazer previsões.

**Exemplos de modelos:**
- **Regressão linear:** busca estabelecer uma reta de tendência central a partir dos dados, da qual se extrapolam previsões.
- **Classificação linear:** semelhante à regressão linear, mas usado para classificar os dados em categorias diferentes.

```python
# O sklearn possui muitas classes, aqui, importamos só a de regressão linear.
from sklearn import linear_model
regressao_linear = linear_model.LinearRegression()
```

### **# Aprendizado**
- Na **regressão linear**, a saída é sempre **um único número**, enquanto a entrada é sempre **um vetor**.
Por exemplo, uma sorveteria pode querer fazer a regressão linear do seu número de clientes (que é a saída); Nesse caso, a entrada poderia ser um *vetor unitário [temperatura]*, mas também pode ser um *vetor de vários elementos: [temperatura, horário, preço]*.
Um **dataset** vai ser uma **lista de vetores de entrada, e uma lista de saída.**

```python
# Exemplo da sorveteria:

X = [[20], [21], [25], [29], [31], [32], [32], [34]]
Y = [4, 2, 7, 15, 16, 14, 15, 20]
plt.scatter(X, Y)
regressao_linear.fit(X, Y)

print(regressao_linear.coef_) # coeficientes de x
print(regressao_linear.intercept_) # termo independente

# Isso foi o "treinamento" do modelo, e agora podemos usá-lo para prever a partir de novos valores.

X2 = [[40], [41], [18]]

fig, ax = plt.subplots()
ax.scatter(X, Y, color="blue", label="Dados Reais")

# Plota a reta de tendência central calculada pelo modelo;
ax.plot(X, regressao_linear.predict(X), color="red", label="Reta de Regressão Linear")
ax.legend()
plt.show()

# O método .score() também funciona na Regressão, calculando o R² (ajuste da reta)
print("R² da Regressão:", regressao_linear.score(X, Y))
```

- Na **classificação linear**, o dataset de entrada é *uma lista de vetores*, e o dataset de saída é *uma lista de categorias*.
Por exemplo, para pinguins: 
- x = [peso, grossura do bico]
- y = 'macho' ou 'fêmea'

```python
from sklearn import svm
classificador = svm.LinearSVC(random_state=42, max_iter=10000)

X_ping = [[3.5, 15], [4.2, 21], [3.1, 14], [4.5, 22]]
y_ping = ['femea', 'macho', 'femea', 'macho']

classificador.fit(X_ping, y_ping)
print("Previsão pinguim novo:", classificador.predict([[3.8, 17]]))

# O método .score(X, y) na classificação calcula a Acurácia (taxa de acertos)
print("Acurácia do Classificador:", classificador.score(X_ping, y_ping))

# Ele compara os valores de saída previstos com os valores de saída reais, e depois calcula a taxa de acertos.
```

### **# Kernels**
- Nem todo dataset pode ser modelado ou classificado por linhas, logo, os kernels servem para mapear o dataset em um espaço que **pode** ser modelado por linhas, viabilizando o machine learning linear nesse espaço, e depois desfazer a transformação.
- Isso pode ser especificado no sklearn com o parâmetro "kernel= na hora de instanciar um modelo.

```python
# Exemplo:
svm.linearSVC(kernel="rbf")
```
